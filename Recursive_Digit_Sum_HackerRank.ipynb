```python
def superDigit(n, k):
    if len(n*k)==1:
        return n
    if len(n*k)>1:
        digit_sum = sum(int(d) for d in n)*k
        return digit_sum
    return superDigit(digit_sum, 1)
```

```python
def superDigit(n, k):
    # Sumar los dígitos de n y multiplicar por k
    digit_sum = sum(int(d) for d in n) * k
    
    # Caso base: si ya es un solo dígito
    if digit_sum < 10:
        return digit_sum
    
    # Llamada recursiva con digit_sum como string
    return superDigit(str(digit_sum), 1)
```

```python
def climbingLeaderboard(ranked, player):
    # Paso 1: Eliminar duplicados y ordenar (por si acaso no está ordenado)
    ranked = sorted(list(set(ranked)), reverse=True)
    n = len(ranked)
    result = []
    
    for score in player:
        # Paso 2: Búsqueda binaria para encontrar la posición
        left, right = 0, n - 1
        position = n  # Por defecto, si es el menor
        
        while left <= right:
            mid = (left + right) // 2
            if ranked[mid] <= score:
                position = mid
                right = mid - 1
            else:
                left = mid + 1
        
        # Paso 3: El rango es position + 1 (Dense Ranking)
        result.append(position + 1)
    
    return result
```

```python
def climbingLeaderboard(ranked, player):
    # Eliminar duplicados y ordenar
    ranked = list(set(ranked))
    ranked.sort()
    n = len(ranked)
    result = []
    i=0
    
    for score in player:      
        while i < n and ranked[i]<=score:
           i = i+1
        result.append(n-i+1)
        
    return result
```
